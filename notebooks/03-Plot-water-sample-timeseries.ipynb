```python
# Import modules
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
# Load data from notebook 1 into a Pandas Dataframe
df = pd.read_csv("./data/966896_v1_ooi_pioneer_nes_ctd_and_water_sampling_data.csv")
df.head() # Print the first 5 rows of the DataFrame
```

```python
# List all unique Target_Asset values
available_assets = df['Target_Asset'].unique()
print("Available Target Assets:", available_assets)
```

```python
# Choose a specific asset to query
asset_to_match = "CNSM"  # Replace with any value from the list above

# Filter the DataFrame for the selected asset and reset index
asset_query = df.query("`Target_Asset` == @asset_to_match").reset_index(drop=True)

# Show the result
asset_query.head()
```

```python
asset_query.columns
```

```python
some_vars = asset_query[["CTD_Depth", "Discrete_Oxygen", "Discrete_Chlorophyll", "Discrete_Nitrate"]]
```

```python
some_vars
```

```python
cast_cmap = mpl.colormaps['Accent'].reversed() # Reversed in attempt to improve colorblind-friendliness
```

```python
time = pd.to_datetime(asset_query["CTD_Bottle_Closure_Time"], yearfirst=True, utc=True).values
```

```python
# Create the subplots
fig, axs = plt.subplots(3, 1, sharex=True, layout="tight")
# Discrete Oxygen
plt.subplot(3,1,1)
plt.scatter(x=time, y=asset_query["Discrete_Oxygen"], c=asset_query["Cast"].astype(float), cmap=cast_cmap, edgecolors="none")
plt.grid()
axs[0].set_ylabel("Discrete Oxygen")
# CTD Fluorescence, I wanted to use Discrete Chlorophyll but one of the values couldn't be parsed as a float ("05/01")
plt.subplot(3,1,2)
plt.scatter(x=time, y=asset_query["CTD_Fluorescence"], c=asset_query["Cast"].astype(float), cmap=cast_cmap, edgecolors="none")
plt.grid()
axs[1].set_ylabel("CTD Fluorescence")
# Discrete Nitrate
plt.subplot(3,1,3)
plt.scatter(x=time, y=asset_query["Discrete_Nitrate"], c=asset_query["Cast"].astype(float), cmap=cast_cmap, edgecolors="none")
plt.grid()
axs[2].set_ylabel("Discrete Nitrate")
axs[2].set_xlabel("Time")
```
